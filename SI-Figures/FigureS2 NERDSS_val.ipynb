```python
def readGillespieSurvP(filename):
    # No estimation of standard error since it is a single experiment
    with open(filename) as f:
        reading = False
        times = []
        survP = []
        numDiss = 'Not Read'
        for line in f:
            if line.strip() == '# START SURVIVAL PROB':
                reading = True
            elif line.strip() == '# END SURVIVAL PROB':
                reading = False
            elif line.strip() == '# Mean Residence Time':
                meanResT = float(f.readline().strip())
            elif line.strip().startswith('# Bootstrap'):
                # Example: "# Bootstrap (100 resamples): 8.2217e-01 ± 6.0352e-03"
                parts = line.split(':', 1)[1].strip().split('±')
                meanResT_BS = float(parts[0])
                stdErr_BS = float(parts[1])
            elif line.strip().startswith('Number of dissociation events:'):
                numDiss = int(line.split(':')[1])
            else:
                if reading:
                    linelist = line.split()
                    times.append(float(linelist[0]))
                    survP.append(float(linelist[1]))
    return np.array(times), np.array(survP), meanResT, numDiss, meanResT_BS, stdErr_BS

def get_survival_prob(resident_times, bins=100):

    # get survival probabilities
    h = np.histogram(resident_times, bins=bins, density=True)
    return h[1], np.append([1], 1 - np.cumsum(h[0]*np.diff(h[1])))

def bootstrapSurvProb(resTime_nerdss, tbins, N_batch=500):
    N_sample = len(resTime_nerdss)
    # print(N_sample, batchSize)
    survprobList = []
    mean_list = [] # for direct mean value
    for _ in range(N_batch):
        randomChoices = np.random.choice(resTime_nerdss, size=N_sample, replace=True)
        mean_list.append(np.mean(randomChoices))
        time, survP = get_survival_prob(randomChoices, tbins)
        survprobList.append(survP)
    return time, survprobList, mean_list
```

```python
pfolder_NERDSS = './Simulations/NERDSS_PNPN/'
pfolder_Gillespie = './Simulations/NERDSS_PNPN/Gillespie/'
```

```python
def readNERDSS_survP_repeats(pfolder, NerdssRepeatIDs, nerdssSubFolder, dt, startT):
    '''read survival probability from association and dissociations'''

    # read residence times from NERDSS kinetics
    resTime_nerdss = []
    resTime_means = []
    for i in NerdssRepeatIDs:
        jobfolder = pfolder_NERDSS+f'/{pfolder}/{nerdssSubFolder}/{i}/'
        resTi, _ = readResT_from_NERDSS(
            jobfolder+'/DATA/assoc_dissoc_time.dat', dt, startT=startT
        )
        resTime_nerdss.append(resTi)
        resTime_means.append(np.mean(resTi))
        
    resTime_nerdss = np.concatenate(resTime_nerdss)

    return resTime_nerdss, resTime_means
    

def compareNerdssWithGillespie(
    dt, GillepieID, folders, NerdssRepeatIDs, startT=10, binwidth=0.05, timeUpperLimit=1000, ifplot=True,
    nerdssSubFolder='nerdss_output',
):
    results = []
    for j, pfolder in zip(GillepieID, folders):
        # read and process survival probability from gillespie
        gllspFile = pfolder_Gillespie+f'/OUTPUTS_Nonly/out_{j}_r0'
        (timeGllsp, survPGllsp, meanRestGllsp, 
         numEvent, meanResT_BS, stdErr_BS) = readGillespieSurvP(gllspFile)
        print(f'Gillespie tau: {meanResT_BS:.5f} +- {stdErr_BS:.6f}')
        
        # read residence times from NERDSS kinetics
        resTime_nerdss, resTime_means = readNERDSS_survP_repeats(
            pfolder, NerdssRepeatIDs, nerdssSubFolder, dt, startT
        )

        # get survival probability from NERDSS residence times
        timeNERDSS, survprobList, BSmeans = bootstrapSurvProb(
            resTime_nerdss, np.arange(0, timeUpperLimit, binwidth)
        )
        survP_mean = np.mean(survprobList, axis=0)
        survP_std = np.std(survprobList, axis=0)

        print(f'{round(np.shape(resTime_nerdss)[0]/48/20)} dissociations / mol found in NERDSS!')
        print(f'NERDSS tau: {np.mean(BSmeans):.5f} +- {np.std(BSmeans):.6f}')

        if ifplot:
            survpFilter = survP_mean>1e-3
            # plot
            # setup figure properties
            fig = plt.figure(figsize=(5,5))
            ax = fig.add_subplot(111)
            plt.tick_params(length=15, width=2)
            for axis in ['top', 'bottom', 'left', 'right']:
                ax.spines[axis].set_linewidth(2)  # change width
            # nerdss result with +- error
            plt.fill_between(
                timeNERDSS[survpFilter], 
                survP_mean[survpFilter] - 1*survP_std[survpFilter], 
                survP_mean[survpFilter] + 1*survP_std[survpFilter], 
                label='NERDSS', alpha=0.5
            )
            # gillespie results
            plt.plot(
                timeGllsp[survPGllsp>1e-3], survPGllsp[survPGllsp>1e-3], 
                label='Gillespie', color='black', lw=1.5
            )   

            plt.semilogy()

            plt.xlim([0, min(timeGllsp[survPGllsp>1e-3][-1], timeNERDSS[survpFilter][-1])])
            plt.ylim([1e-3,1e0])
            if j==3:
                plt.xticks([0, 3, 6, 9, 12, 15, 18], fontsize=22)
            else:
                plt.xticks(fontsize=22)
            
            plt.yticks(fontsize=22)
            
            plt.legend(fontsize=14)
            
            # inset for mean +- std
            inset = ax.inset_axes([0.15, 0.12, 0.25, 0.25])
            inset.errorbar(
                [0, 1],
                [meanResT_BS, np.mean(BSmeans)],
                yerr=[stdErr_BS, np.std(BSmeans)],
                fmt='o', capsize=5, color='black'
            )
            inset.set_title('<τ>', fontsize=14)
            inset.set_xticks([0, 1])
            inset.set_xticklabels(['Gillespie', 'NERDSS'], fontsize=12, rotation=-15)
            inset.tick_params(axis='both', labelsize=10)
            inset.set_xlim([-0.5, 1.5])

            plt.show()

        results.append([
            meanRestGllsp, 
            np.mean(resTime_means), 
            np.std(resTime_means)/np.sqrt(len(resTime_means))
        ])
        
    return np.array(results)
```

```python
NerdssRepeatIDs = [i for i in range(1,49)]
EQUI_mean = []
EQUI_err = []
for job in nerdss_jobs:
    nerdss = nerdssData(NerdssRepeatIDs, pfolder_NERDSS+'/'+job+'/SIMS')
    nerdss.readHist(dataHist='DATA/histogram_complexes_time.dat', ifprint=False)
    mean_list = []
    err_list = []
    for s in species:
        print(s)
        if s not in nerdss.MOLS:
            mean_list.append(0)
            err_list.append(0)
        else:
            windowsize, _, mean,std = find_equi_linear(nerdss.time, nerdss.MOLS[s], ifdraw=False)
            error = np.sqrt(std**2 / windowsize + np.mean(nerdss.MOLS_err[s][-windowsize:])**2)
            mean_list.append(mean)
            err_list.append(error)
    EQUI_mean.append(mean_list)
    EQUI_err.append(err_list)
```
